```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Datos simulados (reemplaza esto con tu lista 'historial')
historial = [
    {'id': 1, 'potencia': 1500, 'fecha': '2023-11-01'},
    {'id': 2, 'potencia': 2200, 'fecha': '2023-11-02'},
    {'id': 3, 'potencia': 1800, 'fecha': '2023-11-03'},
    {'id': 4, 'potencia': 3000, 'fecha': '2023-11-04'},
    {'id': 5, 'potencia': 1950, 'fecha': '2023-11-05'},
    {'id': 6, 'potencia': 2001, 'fecha': '2023-11-06'},
    {'id': 7, 'potencia': 2500, 'fecha': '2023-11-07'}
]

# 2. Conversión a DataFrame
df = pd.DataFrame(historial)

# 3. Creación de la columna 'ACCION'
# Se utiliza una función lambda para aplicar la condición fila por fila
df['ACCION'] = df['potencia'].apply(lambda x: 'VENDER' if x > 2000 else 'ACUMULAR')

# 4. Mostrar las primeras 5 filas
print(df.head())

# 5. Gráfico simple
plt.figure(figsize=(10, 6))
colors = df['ACCION'].map({'VENDER': 'red', 'ACUMULAR': 'green'})
plt.bar(df['id'], df['potencia'], color=colors)
plt.axhline(y=2000, color='blue', linestyle='--', label='Umbral (2000)')
plt.title('Potencia vs Acción')
plt.xlabel('ID')
plt.ylabel('Potencia')
plt.legend()
plt.show()
```

```python
import json
from solcx import compile_standard, install_solc

# 1. Definición del Contrato Solidity (Source Code)
solidity_source = """
// SPDX-License-Identifier: MIT
pragma solidity 0.8.0;

contract EnergyMarket {
    // Mapping de balances: dirección => cantidad
    mapping(address => uint256) public balances;

    // Función vender: suma la cantidad al balance del sender
    function vender(uint256 cantidad) public {
        balances[msg.sender] += cantidad;
    }
}
"""

# 2. Instalación de la versión específica del compilador
# Esto asegura que tengas el binario de solc 0.8.0 descargado
install_solc('0.8.0')

# 3. Compilación usando el formato JSON estándar (Standard JSON Input)
compiled_sol = compile_standard(
    {
        "language": "Solidity",
        "sources": {"EnergyMarket.sol": {"content": solidity_source}},
        "settings": {
            "outputSelection": {
                "*": {
                    "*": ["abi", "metadata", "evm.bytecode", "evm.sourceMap"]
                }
            }
        },
    },
    solc_version="0.8.0",
)

# 4. Extracción de datos
# La estructura del JSON de salida es: contracts -> nombre_archivo -> nombre_contrato
contract_interface = compiled_sol["contracts"]["EnergyMarket.sol"]["EnergyMarket"]

# Guardamos el ABI y el Bytecode en variables
contract_abi = contract_interface["abi"]
contract_bytecode = contract_interface["evm"]["bytecode"]["object"]

# --- Verificación de resultados ---
print("--- ABI (Primeros 100 caracteres) ---")
print(json.dumps(contract_abi)[:100] + "...")

print("\n--- Bytecode (Primeros 100 caracteres) ---")
print(contract_bytecode[:100] + "...")
```

```python
import pandas as pd
from web3 import Web3
from solcx import compile_standard, install_solc
from web3.middleware import buffered_gas_estimate # Import the middleware

# ---------------------------------------------------------
# 1. Configuración de Datos y Compilación (Pasos previos)
# ---------------------------------------------------------

# Crear el DataFrame (Simulando el paso 1)
data = [
    {'id': 1, 'potencia': 1500, 'ACCION': 'ACUMULAR'},
    {'id': 2, 'potencia': 2200, 'ACCION': 'VENDER'},
    {'id': 3, 'potencia': 1800, 'ACCION': 'ACUMULAR'},
    {'id': 4, 'potencia': 3000, 'ACCION': 'VENDER'},
    {'id': 5, 'potencia': 1950, 'ACCION': 'ACUMULAR'}
]
df = pd.DataFrame(data)

# Compilar Contrato (Simulando el paso 2 para obtener ABI/Bytecode)
install_solc('0.8.0')
source = """
// SPDX-License-Identifier: MIT
pragma solidity 0.8.0;
contract EnergyMarket {
    mapping(address => uint256) public balances;
    function vender(uint256 cantidad) public {
        balances[msg.sender] += cantidad;
    }
}
"""
compiled = compile_standard({
    "language": "Solidity",
    "sources": {"EnergyMarket.sol": {"content": source}},
    "settings": {"outputSelection": {"*": {"*": ["abi", "evm.bytecode"]}}}
}, solc_version="0.8.0")

contract_interface = compiled["contracts"]["EnergyMarket.sol"]["EnergyMarket"]
abi = contract_interface["abi"]
bytecode = contract_interface["evm"]["bytecode"]["object"]

# ---------------------------------------------------------
# 2. Configuración de Web3 y eth-tester
# ---------------------------------------------------------

# Conectar a un proveedor de Ethereum en memoria (eth-tester)
w3 = Web3(Web3.EthereumTesterProvider())

# Remove the buffered_gas_estimate_middleware that attempts to estimate gas
# and is not supported by MockBackend. This will prevent the MethodUnavailable error.
# This middleware is not added by default by EthereumTesterProvider when py-evm is not found,
# so attempting to remove it causes a Web3ValueError.
# w3.middleware_onion.remove(buffered_gas_estimate) # REMOVED THIS LINE COMPLETELY

# Verificar conexión
if not w3.is_connected():
    raise Exception("Error al conectar con el proveedor Ethereum")

# Obtener cuenta predeterminada (con fondos de prueba)
cuenta = w3.eth.accounts[0]
print(f"Usando cuenta: {cuenta}\n")

# Set a default gas limit for transactions, as estimation is now disabled.
# A sufficiently high value (e.g., 3,000,000) should be ample for testing.
w3.eth.default_gas = 3000000

# ---------------------------------------------------------
# 3. Despliegue del Contrato
# ---------------------------------------------------------

EnergyMarket = w3.eth.contract(abi=abi, bytecode=bytecode)
tx_hash_deploy = EnergyMarket.constructor().transact({'from': cuenta})
tx_receipt_deploy = w3.eth.wait_for_transaction_receipt(tx_hash_deploy)
contract_address = tx_receipt_deploy.contractAddress

# Instancia del contrato desplegado
market_contract = w3.eth.contract(address=contract_address, abi=abi)
print(f"Contrato desplegado en: {contract_address}\n")

# ---------------------------------------------------------
# 4. Lógica de Negocio: Recorrer DataFrame y Vender
# ---------------------------------------------------------

print("--- Procesando Transacciones ---")

for index, row in df.iterrows():
    potencia = int(row['potencia']) # Asegurar tipo int nativo para Web3
    accion = row['ACCION']

    print(f"ID {row['id']}: Potencia {potencia} -> {accion}")

    if accion == 'VENDER':
        # Llamada a la función 'vender' del contrato
        tx_hash = market_contract.functions.vender(potencia).transact({'from': cuenta})

        # Esperar confirmación (opcional en eth-tester, pero buena práctica)
        w3.eth.wait_for_transaction_receipt(tx_hash)

        print(f"   [TX ENVIADA] Hash: {tx_hash.hex()}")
    else:
        print("   [OMITIDO] No se cumple condición de venta.")

# ---------------------------------------------------------
# 5. Verificación Final
# ---------------------------------------------------------

# Consultar balance final en el mapping del contrato
saldo_final = market_contract.functions.balances(cuenta).call()

print("\n--- Resumen ---")
print(f"Saldo acumulado en el contrato para {cuenta}: {saldo_final}")
```
